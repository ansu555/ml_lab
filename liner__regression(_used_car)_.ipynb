**Helper Functions**

```python
# 📊 Load and clean the dataset
import pandas as pd

df = pd.read_csv("Used_Car_Dataset - Used_Car_Dataset.csv")
df.dropna(inplace=True)

# Select numerical columns only
df = df[['Year', 'Mileage_km', 'Engine_Size_cc', 'Num_Owners', 'Fuel_Efficiency_kmpl', 'Price_INR']]
df.head()
```

**General Gradient Descent Function**


```python
import numpy as np
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import r2_score
import matplotlib.pyplot as plt

# Gradient Descent Function
def gradient_descent(X, y, alpha=0.01, epochs=10000, tol=1e-6):
    m, n = X.shape
    theta = np.zeros((n, 1))
    prev_cost = float('inf')
    for i in range(epochs):
        pred = X @ theta
        error = pred - y
        cost = (1 / (2 * m)) * np.sum(error ** 2)
        if abs(prev_cost - cost) < tol:
            print(f"✅ Converged at iteration {i}")
            break
        prev_cost = cost
        grad = (1 / m) * X.T @ error
        theta -= alpha * grad
    return theta, cost

# R² Score Evaluation
def evaluate_model(X, y, theta):
    pred = X @ theta
    score = r2_score(y, pred)
    print(f"🔍 R² Score: {score:.4f} ({score * 100:.2f}% accuracy)")
    return pred, score
```

```python
# Select input and target
x_uni = df[['Engine_Size_cc']].values
y_uni = df['Price_INR'].values.reshape(-1, 1)

# Normalize X and add bias term
scaler_uni = StandardScaler()
x_uni_scaled = scaler_uni.fit_transform(x_uni)
x_uni_scaled = np.c_[np.ones((x_uni_scaled.shape[0], 1)), x_uni_scaled]

# Run Gradient Descent
theta_uni, cost_uni = gradient_descent(x_uni_scaled, y_uni, alpha=0.01, epochs=10000, tol=1e-6)

# Evaluate
pred_uni, r2_uni = evaluate_model(x_uni_scaled, y_uni, theta_uni)
print(f"🧠 Hypothesis: h(x) = {theta_uni[0][0]:.2f} + {theta_uni[1][0]:.2f} * Engine_Size_cc")
```

**Accuracy (R²) and Hypothesis Display**

```python
# Plot the regression line
plt.scatter(x_uni, y_uni, color='blue', label='Actual')
x_pred = scaler_uni.transform(x_uni)
x_pred = np.c_[np.ones(x_pred.shape[0]), x_pred]
plt.plot(x_uni, x_pred @ theta_uni, color='red', label='Predicted')
plt.xlabel("Engine Size (cc)")
plt.ylabel("Price (INR)")
plt.title("Univariate Linear Regression - Used Car Price")
plt.legend()
plt.show()

```

**Step 4: Run Multivariate Linear Regression**

```python
# MULTIVARIATE LINEAR REGRESSION
features = ['Year', 'Mileage_km', 'Engine_Size_cc', 'Num_Owners', 'Fuel_Efficiency_kmpl']
df_mv = df[features + ['Price_INR']].dropna()

X_mv = df_mv[features].values
y_mv = df_mv['Price_INR'].values.reshape(-1, 1)

# Normalize
sc_mv = StandardScaler()
X_scaled = sc_mv.fit_transform(X_mv)
X_scaled = np.c_[np.ones((X_scaled.shape[0], 1)), X_scaled]

# Train
theta_mv, cost_mv = gradient_descent(X_scaled, y_mv, alpha=0.01, epochs=10000)
pred_mv, r2_mv = evaluate_model(X_scaled, y_mv, theta_mv)

# Display model
feature_names = ['bias'] + features
terms = [f"{theta_mv[i][0]:+.2f} * {name}" for i, name in enumerate(feature_names)]
print("🧠 Multivariate Hypothesis:\nh(x) =", " ".join(terms))
```

```python
alphas = [0.0001, 0.001, 0.01, 0.1]
best_r2 = -1
best_alpha = None

for a in alphas:
    theta_tune, _ = gradient_descent(X_scaled, y_mv, alpha=a)
    _, r2_tune = evaluate_model(X_scaled, y_mv, theta_tune)
    if r2_tune > best_r2:
        best_r2 = r2_tune
        best_alpha = a

print(f"🔧 Best learning rate: {best_alpha} with R²: {best_r2:.4f}")
```
